## 6. Classes abstraites

### 6.1 Imposer un contrat

**Objectif** : Obliger chaque forme à savoir calculer son aire.

**À faire** :
1. `Forme` hérite de `ABC` et déclare `aire` abstraite (`@abstractmethod`, corps `...`).
2. `Carre.aire` renvoie `cote ** 2`, `Disque.aire` renvoie `math.pi * rayon ** 2`.
3. La méthode concrète `decrire` est déjà écrite : ne la touche pas.

**Résultat attendu** :
```
Carre d'aire 4.0
Disque d'aire 3.1
```

📖 Cours [05-heritage-polymorphisme-composition.ipynb](05-heritage-polymorphisme-composition.ipynb), section « 6. Classes abstraites », cellules « ABC + @abstractmethod » et « Impossible d'instancier le contrat ».


````python
from abc import ABC, abstractmethod
import math

class Forme(...):          # hérite de ABC
    ...                    # décorateur qui rend aire() abstraite
    def aire(self):
        ...                # pas de corps : laisse ... tel quel

    def decrire(self):
        return f"{type(self).__name__} d'aire {self.aire():.1f}"

class Carre(Forme):
    def __init__(self, cote):
        self.cote = cote

    def aire(self):
        ...                # cote ** 2

class Disque(Forme):
    def __init__(self, rayon):
        self.rayon = rayon

    def aire(self):
        ...                # math.pi * rayon ** 2

for f in [Carre(2), Disque(1)]:
    print(f.decrire())
````

````python
# Vérification
assert Carre(3).aire() == 9 and abs(Disque(2).aire() - 4 * math.pi) < 1e-9
try:
    Forme()
except TypeError:
    pass
else:
    raise AssertionError("Forme doit être abstraite : impossible à instancier")
assert Carre(2).decrire() == "Carre d'aire 4.0"
print("OK")
````

````python
class Moteur:
    def demarrer(self):
        return "vroum"

class Voiture(Moteur):      # ❌ à corriger : Voiture ne doit plus hériter de Moteur
    pass                    # ajoute __init__ (crée self.moteur = Moteur()) et demarrer() qui délègue au moteur

v = Voiture()
print(v.demarrer(), issubclass(Voiture, Moteur))
````
