**Objectif** : Donner à chaque objet sa propre liste.

**À faire** :
1. La liste `notes` est un attribut de classe : toutes les élèves partagent la même.
2. Déplace-la dans `__init__` pour qu'elle soit propre à chaque objet.


````python
# Un bug à corriger : la liste notes est partagée par toutes les élèves
class Eleve:
    notes = []

    def __init__(self, nom):
        self.nom = nom

    def ajouter_note(self, note):
        self.notes.append(note)

a, b = Eleve("Ana"), Eleve("Bo")
a.ajouter_note(15)
print(a.notes, b.notes)
````
