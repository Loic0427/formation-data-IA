## 1. Pourquoi des objets ?

Un programme manipule des **données** (un compte : titulaire, solde) et des **actions** sur ces données (déposer, retirer). En procédural, les deux sont séparés. La programmation orientée objet (POO) les **regroupe dans un objet**.

| Terme | Définition | Dans le code |
|---|---|---|
| Classe | le plan, le modèle | `class CompteBancaire:` |
| Objet, instance | un exemplaire concret fabriqué à partir du plan | `compte = CompteBancaire("Alice")` |
| Attribut | une donnée portée par l'objet : son **état** | `compte.solde` |
| Méthode | une fonction portée par l'objet : son **comportement** | `compte.deposer(50)` |
| Constructeur | la méthode appelée automatiquement à la création | `def __init__(self, ...)` |
| `self` | l'objet lui-même, vu de l'intérieur de la classe | `self.solde` |


```python
# Version procédurale : les données (un dict) et les fonctions sont séparées
compte = {"titulaire": "Alice", "solde": 100.0}

def deposer(compte, montant):
    compte["solde"] += montant

deposer(compte, 50)
print(compte)
```

```python
# Version objet : données et actions regroupées dans une classe
class CompteBancaire:
    def __init__(self, titulaire, solde=0.0):
        self.titulaire = titulaire
        self.solde = solde

    def deposer(self, montant):
        self.solde += montant

compte = CompteBancaire("Alice", 100.0)
compte.deposer(50)
print(compte.titulaire, compte.solde)
```

## 2. Classe et objets

`class Nom:` définit le plan (nom en `PascalCase`). `Nom(...)` fabrique un objet : c'est l'**instanciation**. On peut créer autant d'objets qu'on veut, chacun avec son propre état.


```python
# La classe la plus simple possible, et deux objets créés à partir d'elle
class Chien:
    pass

rex = Chien()
medor = Chien()
print(type(rex))
print(isinstance(rex, Chien), rex is medor)
```

```python
# Chaque objet a ses propres attributs : modifier l'un ne touche pas l'autre
a = CompteBancaire("Alice", 100)
b = CompteBancaire("Bob", 100)
a.deposer(500)
print(a.solde, b.solde)
```

```python
# Deux objets au même état restent deux objets distincts ; une affectation ne copie pas
c1 = CompteBancaire("Alice", 100)
c2 = CompteBancaire("Alice", 100)
print(c1 is c2)                # False : deux objets différents

alias = c1                     # alias désigne LE MÊME objet que c1
alias.deposer(10)
print(c1.solde, alias is c1)
```

## 3. `__init__` et `self`

`__init__` s'exécute automatiquement à la création et reçoit les arguments passés entre parenthèses. `self` est le premier paramètre de toute méthode d'instance : Python y place l'objet en cours. `compte.deposer(50)` est un raccourci pour `CompteBancaire.deposer(compte, 50)`.


```python
# __init__ range les arguments dans self ; on peut aussi y créer des attributs sans paramètre
class Livre:
    def __init__(self, titre, auteur, annee=None):
        self.titre = titre
        self.auteur = auteur
        self.annee = annee
        self.disponible = True         # état initial, pas un paramètre

livre = Livre("Dune", "Herbert", 1965)
print(livre.titre, livre.annee, livre.disponible)
```

```python
# self n'a rien de magique : c'est l'objet passé en premier argument
class Compteur:
    def __init__(self):
        self.valeur = 0

    def incrementer(self):
        self.valeur += 1

c = Compteur()
c.incrementer()                # forme habituelle
Compteur.incrementer(c)        # forme équivalente : self = c
print(c.valeur)
```

```python
# vars() montre l'état d'un objet : ses attributs d'instance
print(vars(livre))
```

## 4. Méthodes

Une méthode lit ou modifie l'état via `self`, et peut appeler une autre méthode avec `self.autre()`. Comme une fonction, elle renvoie une valeur avec `return`, sinon `None`.


```python
# Trois rôles de méthode : lire l'état, le modifier, combiner
class Rectangle:
    def __init__(self, largeur, hauteur):
        self.largeur = largeur
        self.hauteur = hauteur

    def aire(self):                        # lit l'état
        return self.largeur * self.hauteur

    def agrandir(self, facteur):           # modifie l'état
        self.largeur *= facteur
        self.hauteur *= facteur

    def est_carre(self):                   # combine
        return self.largeur == self.hauteur

r = Rectangle(2, 3)
r.agrandir(2)
print(r.aire(), r.est_carre())
```

```python
# Une méthode peut refuser une action et le signaler par sa valeur de retour
class CompteBancaire:
    def __init__(self, titulaire, solde=0.0):
        self.titulaire = titulaire
        self.solde = solde

    def retirer(self, montant):
        if montant > self.solde:
            return False
        self.solde -= montant
        return True

compte = CompteBancaire("Alice", 100)
print(compte.retirer(150), compte.retirer(30), compte.solde)
```

## 5. Afficher un objet : `__repr__` et `__str__`

Par défaut, `print(objet)` affiche `<__main__.Livre object at 0x...>`. Deux méthodes spéciales corrigent ça : `__repr__` donne l'affichage **technique** (console, contenu d'une liste, debug) et `__str__` l'affichage **lisible** (`print`, f-string). Définir au moins `__repr__` ; sans `__str__`, `print` utilise `__repr__`.


```python
# Sans __repr__, l'affichage est inutilisable
print(Livre("Dune", "Herbert"))
```

```python
# __repr__ : non ambigu, idéalement le code qui recrée l'objet ; __str__ : pour l'humain
class Livre:
    def __init__(self, titre, auteur):
        self.titre = titre
        self.auteur = auteur

    def __repr__(self):
        return f"Livre({self.titre!r}, {self.auteur!r})"

    def __str__(self):
        return f"{self.titre}, de {self.auteur}"

livre = Livre("Dune", "Herbert")
print(livre)                   # __str__
print([livre])                 # __repr__ dans un conteneur
print(repr(livre))
```

## 6. Attributs de classe

Un attribut défini dans la classe, **hors** de `__init__`, est **partagé** par toutes les instances : constante, valeur par défaut, compteur. On le lit via `self.X` ou `Classe.X`, on le modifie via `Classe.X`.


```python
# Attribut de classe : une seule valeur pour tous les objets
class CompteBancaire:
    TAUX_INTERET = 0.02            # constante partagée
    nb_comptes = 0                 # compteur partagé

    def __init__(self, titulaire, solde=0.0):
        self.titulaire = titulaire
        self.solde = solde
        CompteBancaire.nb_comptes += 1

    def interets(self):
        return self.solde * self.TAUX_INTERET

a = CompteBancaire("Alice", 1000)
b = CompteBancaire("Bob", 500)
print(a.interets(), CompteBancaire.nb_comptes)
```

```python
# Piège : affecter via l'instance crée un attribut d'INSTANCE qui masque celui de la classe
a.TAUX_INTERET = 0.05
print(a.TAUX_INTERET, b.TAUX_INTERET, CompteBancaire.TAUX_INTERET)
print(vars(a))                 # TAUX_INTERET est maintenant dans l'état de a
```

## 7. Méthodes de classe et méthodes statiques

`@classmethod` reçoit la **classe** (`cls`) au lieu de l'objet : sert surtout de **constructeur alternatif** (depuis une chaîne, un dict, un fichier). `@staticmethod` ne reçoit ni objet ni classe : une simple fonction rangée dans la classe parce qu'elle lui est liée.


```python
# @classmethod construit l'objet autrement ; @staticmethod est une fonction utilitaire
class Date:
    def __init__(self, jour, mois, annee):
        self.jour, self.mois, self.annee = jour, mois, annee

    @classmethod
    def depuis_iso(cls, texte):            # "2026-10-18"
        annee, mois, jour = texte.split("-")
        return cls(int(jour), int(mois), int(annee))

    @staticmethod
    def est_bissextile(annee):             # ni self ni cls
        return annee % 4 == 0 and (annee % 100 != 0 or annee % 400 == 0)

    def __repr__(self):
        return f"Date({self.jour}, {self.mois}, {self.annee})"

d = Date.depuis_iso("2026-10-18")
print(d, Date.est_bissextile(2024), Date.est_bissextile(2026))
```

## 8. Tout est objet

En Python, `42`, `"abc"`, une liste, une fonction sont des objets. Les méthodes des types natifs (`texte.upper()`, `liste.append()`) suivent exactement la logique de celles que tu écris.


```python
# Les types natifs sont des classes : même syntaxe objet.methode()
print(type(42), type("abc"), type([]))
print("abc".upper(), isinstance(42, int))
```

## ⚠️ Pièges

- Oublier `self` dans la signature : `TypeError: ... takes 0 positional arguments but 1 was given`.
- Oublier `self.` devant un attribut : on crée une variable locale qui disparaît à la fin de la méthode.
- Appeler une méthode d'instance sur la classe sans objet : `Chien.aboyer()`.
- `__init__` ne doit rien renvoyer : pas de `return valeur`.
- Attribut de classe **mutable** (`notes = []`) : partagé entre toutes les instances. Les listes et dicts s'initialisent dans `__init__`.
- Confondre `objet.methode` (la méthode elle-même) et `objet.methode()` (son appel).


```python
# BUG classique : self oublié dans la signature
class Personne:
    def __init__(self, nom):
        self.nom = nom

    def saluer():                  # il manque self
        return "bonjour"

try:
    Personne("Alice").saluer()
except TypeError as e:
    print("TypeError :", e)
```

```python
# BUG classique : liste en attribut de classe, partagée par tous les élèves
class Eleve:
    notes = []                     # ❌ une seule liste pour toute la classe

    def __init__(self, nom):
        self.nom = nom

a, b = Eleve("Ana"), Eleve("Bo")
a.notes.append(15)
print(b.notes)                     # [15] : Bo a la note d'Ana
```

```python
# Correction : la liste est créée dans __init__, une par objet
class Eleve:
    def __init__(self, nom):
        self.nom = nom
        self.notes = []            # ✅ propre à chaque objet

a, b = Eleve("Ana"), Eleve("Bo")
a.notes.append(15)
print(a.notes, b.notes)
```
