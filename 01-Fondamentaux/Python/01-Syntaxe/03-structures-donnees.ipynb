# 02 - Structures de données

| Type | Ordonné | Mutable | Doublons | Syntaxe | Usage |
|---|---|---|---|---|---|
| `str` | oui | non | oui | `"abc"` | texte |
| `list` | oui | oui | oui | `[1, 2]` | collection modifiable |
| `tuple` | oui | non | oui | `(1, 2)` | enregistrement fixe, clé de dict |
| `dict` | oui (insertion) | oui | clés uniques | `{"a": 1}` | association clé → valeur |
| `set` | non | oui | non | `{1, 2}` | unicité, opérations ensemblistes |

Les **séquences** (`str`, `list`, `tuple`) partagent l'indexation, le slicing, `len()`, `in`, `+`, `*`.

## 1. Indexation et slicing (commun aux séquences)

`seq[i]` (négatif = depuis la fin), `seq[start:stop:step]` (stop exclu).

```python
# Indexation (positive et négative) et slicing [début:fin:pas]
s = "Python"
print(s[0], s[-1])
print(s[1:4], s[:2], s[2:])
print(s[::2], s[::-1])
```

```python
# Opérations communes aux séquences : longueur, appartenance, répétition, concaténation
print(len(s), "th" in s, s * 2, s + "3")
```

```python
# Les slices tolèrent les bornes hors limites, pas l'indexation
print(s[2:100])
try:
    s[100]
except IndexError as e:
    print("IndexError :", e)
```

## 2. Strings

Immutables : chaque méthode **retourne une nouvelle chaîne**.

```python
# Nettoyage et casse
s = "  Bonjour le Monde  "
print(s.strip(), s.lower(), s.upper(), s.title(), sep=" | ")
```

```python
# str -> list avec split(), list -> str avec join()
print(s.strip().split(" "))
print("-".join(["a", "b", "c"]))
```

```python
# Remplacer, chercher, compter
print(s.replace("Monde", "Python").strip())
print(s.find("le"), s.count("o"), s.strip().startswith("Bon"))
```

```python
# Tests sur le contenu
print("42".isdigit(), "abc".isalpha(), "   ".isspace())
```

```python
# Guillemets simples ou doubles, chaîne multi-lignes, raw string, échappement
citation = 'Il a dit "bonjour"'
multi = """ligne 1
ligne 2"""
chemin = r"C:\Users\data"      # r = pas d'interprétation des \
print(citation, multi, chemin, "tab\tici", sep="\n")
```

```python
# Immutabilité : on ne modifie pas une str, on en crée une nouvelle
mot = "python"
try:
    mot[0] = "P"
except TypeError as e:
    print("TypeError :", e)
mot = "P" + mot[1:]
print(mot)
```

## 3. Listes

Collection ordonnée et mutable. Les méthodes de modification agissent **en place** et retournent `None`.

```python
# Ajouter : append (1 élément), extend (plusieurs), insert (à une position)
fruits = ["pomme", "banane"]
fruits.append("kiwi")
fruits.extend(["mangue", "cerise"])
fruits.insert(0, "fraise")
print(fruits)
```

```python
# Retirer : remove (par valeur), pop (par index, retourne l'élément), del (par index)
fruits.remove("banane")
dernier = fruits.pop()
premier = fruits.pop(0)
del fruits[0]
print(fruits, dernier, premier)
```

```python
# Chercher : index, count, in
print(fruits.index("mangue"), fruits.count("kiwi"), "kiwi" in fruits)
```

```python
# sorted() crée une nouvelle liste, sort() trie en place
nombres = [3, 1, 4, 1, 5, 9, 2]
print(sorted(nombres))
print(nombres)
nombres.sort(reverse=True)
print(nombres)
```

```python
# Trier selon une clé : insensible à la casse, par longueur
mots = ["banane", "Kiwi", "pomme"]
print(sorted(mots, key=str.lower))
print(sorted(mots, key=len))
```

```python
# Agrégats natifs
print(min(nombres), max(nombres), sum(nombres))
```

```python
# Listes imbriquées
matrice = [[1, 2, 3], [4, 5, 6]]
print(matrice[1][2])
```

```python
# Affectation = même objet (b suit a entièrement)
a = [1, [2, 3]]
b = a
a[0] = 99
print(a, b)
```

```python
# copy() = copie superficielle : la sous-liste reste partagée
a = [1, [2, 3]]
c = a.copy()
a[0] = 99
a[1].append(4)
print(a)
print(c)
```

```python
# deepcopy() = copie profonde : totalement indépendante
import copy
a = [1, [2, 3]]
d = copy.deepcopy(a)
a[1].append(4)
print(a)
print(d)
```

## 4. Tuples

Comme une liste mais **immutable**. Utilisé pour les enregistrements fixes, les retours multiples et les clés de dict.

```python
# La virgule fait le tuple, pas les parenthèses
point = (3, 4)
seul = (5,)
print(type(point), type(seul), type((5)))
```

```python
# Unpacking, avec * pour récupérer "le reste"
x, y = point
premier, *milieu, dernier = [1, 2, 3, 4, 5]
print(x, y)
print(premier, milieu, dernier)
```

```python
# Une fonction qui retourne plusieurs valeurs retourne un tuple
def min_max(valeurs):
    return min(valeurs), max(valeurs)

lo, hi = min_max([4, 8, 1])
print(lo, hi)
```

```python
# Les deux seules méthodes d'un tuple
print(point.count(3), point.index(4))
```

```python
from collections import namedtuple
Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)
print(p.x, p[1], p._asdict())
```

## 5. Dictionnaires

Association clé → valeur. Clés : tout objet **hashable** (str, int, tuple... pas de liste). Ordre d'insertion conservé.

```python
# Créer, ajouter, lire
personne = {"nom": "Alice", "age": 30}
personne["ville"] = "Paris"
print(personne)
print(personne["nom"])
```

```python
# get() : pas de KeyError si la clé est absente
print(personne.get("email"))
print(personne.get("email", "n/a"))
```

```python
# Supprimer : del (par clé), pop (retourne la valeur) ; tester avec in
del personne["ville"]
age = personne.pop("age")
print(personne, age, "nom" in personne)
```

```python
# Les trois vues : keys, values, items
notes = {"maths": 15, "python": 18}
print(list(notes.keys()))
print(list(notes.values()))
print(list(notes.items()))
```

```python
# Parcourir clé et valeur en même temps
for matiere, note in notes.items():
    print(matiere, note)
```

```python
# Fusionner : update() en place, | crée un nouveau dict (3.9+)
notes.update({"sql": 12, "maths": 16})
print(notes)
fusion = notes | {"stats": 14}
print(fusion)
```

```python
# setdefault() : crée la clé seulement si elle est absente
notes.setdefault("git", 10)
notes.setdefault("maths", 0)
print(notes)
```

```python
# Compter sans KeyError avec defaultdict
from collections import defaultdict
compteur = defaultdict(int)
for lettre in "abracadabra":
    compteur[lettre] += 1
print(dict(compteur))
```

```python
# Counter fait la même chose en une ligne
from collections import Counter
print(Counter("abracadabra"))
print(Counter("abracadabra").most_common(2))
```

```python
# Imbriqué + accès sécurisé
config = {"db": {"host": "localhost", "port": 5432}}
print(config["db"]["port"])
print(config.get("cache", {}).get("ttl", 60))
```

## 6. Sets

Éléments **uniques**, non ordonnés, hashables. `{}` vide est un dict : utiliser `set()`.

```python
# Pas de doublons ; attention {} est un dict vide, set() est le set vide
a = {1, 2, 3, 3}
b = set([3, 4, 5])
print(a, b)
print(type({}), type(set()))
```

```python
# Union, intersection, différence, différence symétrique
print(a | b, a & b, a - b, a ^ b)
```

```python
# Inclusion et disjonction
print({1, 2} <= a, a.isdisjoint({9}))
```

```python
# Ajouter / retirer : discard est silencieux, remove lève KeyError si absent
a.add(10)
a.discard(99)
a.remove(10)
print(a)
```

```python
# Dédoublonner en gardant l'ordre
print(list(dict.fromkeys([3, 1, 3, 2])))
```

```python
# frozenset : un set immutable, donc hashable
print(frozenset([1, 2]))
```

## ⚠️ Pièges

- `b = a` ne copie pas une liste ou un dict : les deux noms pointent vers le même objet.
- `liste.sort()` et `liste.append()` retournent `None` : ne pas écrire `liste = liste.sort()`.
- `(5)` est un `int`, `(5,)` est un tuple.
- Un tuple est immutable, mais une liste qu'il contient reste modifiable.
- `{}` crée un dict vide, pas un set.
- Ne jamais modifier une liste ou un dict pendant qu'on itère dessus : itérer sur une copie (`for x in liste[:]`).
- Une liste ne peut pas être clé de dict ni élément de set (`TypeError: unhashable type`).
