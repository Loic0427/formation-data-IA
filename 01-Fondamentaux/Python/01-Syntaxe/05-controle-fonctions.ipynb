## 1. Conditions

Comparaison : `== != < <= > >=` ; logique : `and or not` ; appartenance : `in` ; identité : `is`.
Un bloc commence par `:` et est délimité par l'indentation.

```python
# if / elif / else
note = 14
if note >= 16:
    mention = "très bien"
elif note >= 12:
    mention = "bien"
else:
    mention = "passable"
print(mention)
```

```python
# Comparaison chaînée
print(0 < note < 20)
```

```python
# Expression ternaire
mention = "admis" if note >= 10 else "refusé"
print(mention)
```

```python
# Truthy / falsy : on teste directement la valeur
panier = []
if not panier:
    print("panier vide")
```

```python
# is pour None (identité), == pour les valeurs (égalité)
x = None
print(x is None)
print([] == [], [] is [])
```

```python
# Évaluation paresseuse : le 2e opérande n'est évalué que si nécessaire
nom = ""
print(nom or "anonyme")
```

```python
# match-case (3.10+) : pattern matching structurel
def traiter(commande):
    match commande.split():
        case ["go", direction]:
            return f"avance vers {direction}"
        case ["stop"]:
            return "arrêt"
        case ["pick", *objets] if objets:
            return f"ramasse {', '.join(objets)}"
        case _:
            return "commande inconnue"

for c in ["go nord", "stop", "pick clé lampe", "danse"]:
    print(traiter(c))
```

## 2. Boucles

`for` parcourt un itérable ; `while` boucle tant qu'une condition est vraie. `break` sort, `continue` passe au tour suivant, `else` s'exécute si la boucle n'a pas été interrompue par `break`.

```python
# range(start, stop, step) : stop est exclu
for i in range(1, 10, 3):
    print(i, end=" ")
print()
```

```python
# enumerate() : index + valeur
fruits = ["pomme", "kiwi", "mangue"]
for i, fruit in enumerate(fruits, start=1):
    print(i, fruit)
```

```python
# zip() : itération parallèle sur plusieurs séquences
prix = [1.2, 2.5, 3.0]
for fruit, p in zip(fruits, prix):
    print(f"{fruit}: {p} €")
```

```python
# Parcourir un dict
notes = {"maths": 15, "python": 18}
for matiere, note in notes.items():
    print(matiere, note)
```

```python
# while, continue (saute le tour) et break (sort de la boucle)
n = 0
while True:
    n += 1
    if n % 2 == 0:
        continue
    if n > 7:
        break
    print(n, end=" ")
print()
```

```python
# for ... else : le else s'exécute seulement si la boucle n'a pas été interrompue par break
for candidat in [4, 6, 9, 8]:
    if candidat % 2:
        print("impair trouvé :", candidat)
        break
else:
    print("aucun impair")
```

```python
# Boucles imbriquées
for ligne in range(1, 4):
    print(" ".join(f"{ligne * col:2d}" for col in range(1, 4)))
```

## 3. Fonctions

`def nom(paramètres):` + docstring + corps + `return` (sinon `None`). Arguments positionnels, nommés, par défaut, variables (`*args`, `**kwargs`).

```python
# Définition avec docstring et valeur par défaut
def aire_rectangle(largeur, hauteur=1.0):
    """Retourne l'aire d'un rectangle.

    Args:
        largeur (float): largeur en mètres.
        hauteur (float): hauteur en mètres, 1.0 par défaut.
    Returns:
        float: l'aire.
    """
    return largeur * hauteur
```

```python
# Appel positionnel, avec défaut, par mots-clés
print(aire_rectangle(3, 2))
print(aire_rectangle(3))
print(aire_rectangle(hauteur=4, largeur=2))
```

```python
# La docstring est lisible avec help()
help(aire_rectangle)
```

```python
# *args -> tuple, **kwargs -> dict
def stats(*valeurs, **options):
    """*valeurs -> tuple ; **options -> dict."""
    total = sum(valeurs)
    if options.get("moyenne"):
        return total / len(valeurs)
    return total

print(stats(1, 2, 3))
print(stats(1, 2, 3, moyenne=True))
```

```python
# Dépaquetage à l'appel : * pour une séquence, ** pour un dict
args = [4, 5, 6]
kwargs = {"moyenne": True}
print(stats(*args, **kwargs))
```

```python
# / : paramètres positionnels uniquement ; * : paramètres nommés uniquement
def point(x, y, /, *, label="p"):
    return f"{label}({x}, {y})"

print(point(1, 2, label="A"))
```

```python
# Portée LEGB : Local > Enclosing > Global > Builtins
# global : modifier une variable du module depuis une fonction (sinon UnboundLocalError)
compteur = 0

def incrementer():
    global compteur
    compteur += 1

incrementer()
incrementer()
print(compteur)
```

```python
# nonlocal : modifier la variable de la fonction englobante (closure)
def fabrique_compteur():
    n = 0
    def suivant():
        nonlocal n
        n += 1
        return n
    return suivant

c = fabrique_compteur()
print(c(), c(), c())
```

```python
# Les fonctions sont des objets : on peut les stocker dans un dict
def carre(x):
    return x * x

operations = {"carre": carre, "double": lambda x: 2 * x}   # lambda : fonction anonyme d'une expression
print(operations["carre"](5), operations["double"](5))
```

```python
# ... et les passer en paramètre
def appliquer(fonction, valeurs):
    return [fonction(v) for v in valeurs]

print(appliquer(carre, [1, 2, 3]))
```

```python
# lambda comme clé de tri
print(sorted(["banane", "kiwi", "pomme"], key=lambda m: len(m)))
```

```python
# map() et filter() : fonctions d'ordre supérieur natives
print(list(map(carre, [1, 2, 3])))
print(list(filter(lambda x: x % 2, [1, 2, 3, 4])))
```

## 4. Comprehensions

`[expr for x in iterable if condition]` : construit une liste en une ligne. Même syntaxe pour dict `{k: v ...}`, set `{x ...}` et générateur `(x ...)`.

```python
# List comprehension avec filtre
nombres = range(10)
print([n ** 2 for n in nombres if n % 2 == 0])
```

```python
# if-else dans l'expression (ternaire), avant le for
print(["pair" if n % 2 == 0 else "impair" for n in range(4)])
```

```python
# Dict comprehension et set comprehension
print({mot: len(mot) for mot in ["python", "sql"]})
print({n % 3 for n in nombres})
```

```python
# Expression génératrice : rien n'est stocké en mémoire
print(sum(n ** 2 for n in nombres))
```

```python
# Double for : aplatir une matrice
matrice = [[1, 2], [3, 4]]
print([x for ligne in matrice for x in ligne])
```

## ⚠️ Pièges

- `=` affecte, `==` compare.
- `range(5)` va de 0 à 4. `range(1, n)` exclut `n`.
- Modifier une liste en l'itérant saute des éléments : itérer sur une copie ou construire une nouvelle liste.
- **Argument par défaut mutable** : `def f(items=[])` partage la même liste entre tous les appels. Utiliser `items=None` puis `items = items or []`.
- Oublier `return` : la fonction renvoie `None`.
- Une comprehension à plus de deux niveaux ou avec effets de bord : préférer une boucle lisible.

```python
# BUG classique : une valeur par défaut mutable est partagée entre les appels
def ajouter(valeur, items=[]):
    items.append(valeur)
    return items

print(ajouter(1), ajouter(2))        # [1, 2] [1, 2]
```

```python
# Correction : None par défaut, puis créer la liste dans la fonction
def ajouter_ok(valeur, items=None):
    items = items if items is not None else []
    items.append(valeur)
    return items

print(ajouter_ok(1), ajouter_ok(2))
```
