## 2. Conventions (PEP 8)

| Élément | Convention | Exemple |
|---|---|---|
| variables, fonctions | `snake_case` | `prix_total`, `calculer_tva()` |
| constantes | `UPPER_CASE` | `TAUX_TVA = 0.2` |
| classes | `PascalCase` | `CompteBancaire` |
| indentation | 4 espaces (jamais de tabulation) | |
| longueur de ligne | 79 à 88 caractères | |

Règles d'un identifiant : lettres, chiffres, `_` ; ne commence pas par un chiffre ; sensible à la casse ; pas un mot réservé (`if`, `class`, `def`...).
Un commentaire (`#`) explique le **pourquoi**, pas le quoi.

```python
# Les mots réservés : interdits comme noms de variables
import keyword
print(keyword.kwlist)
```

```python
# Constante en UPPER_CASE, variables en snake_case, espaces autour des opérateurs
TAUX_TVA = 0.2
prix_ht = 100
prix_ttc = prix_ht * (1 + TAUX_TVA)
print(prix_ttc)
```

## 3. Variables et typage dynamique

Une variable est une **étiquette** posée sur un objet. Pas de déclaration de type : le type vient de la valeur.

```python
# Typage dynamique : la même étiquette peut pointer vers un int puis une str
x = 42
print(type(x))
x = "quarante"
print(type(x))
```

```python
# Affectation multiple
a, b = 1, 2
print(f"a = {a}, b = {b}")
```

```python
# Échange sans variable temporaire
a, b = b, a
print(f"a = {a}, b = {b}")
```

```python
# Affectation augmentée (+=, -=, *=, /=, //=, %=, **=)
compteur = 0
compteur += 1
print(compteur)
```

## 4. Types de base

| Type | Exemples | Note |
|---|---|---|
| `int` | `42`, `-7`, `1_000_000` | taille illimitée |
| `float` | `3.14`, `1e-3` | 64 bits, précision limitée |
| `bool` | `True`, `False` | sous-type de `int` |
| `str` | `"texte"` | voir notebook 03 |
| `NoneType` | `None` | absence de valeur |

```python
# Division réelle, division entière, modulo, puissance
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
```

```python
# Fonctions natives utiles sur les nombres
print(round(3.14159, 2), abs(-5), divmod(7, 2))
```

```python
# Précision des flottants : 0.1 + 0.2 n'est pas exactement 0.3
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

```python
# Comparer des flottants avec une tolérance
print(abs((0.1 + 0.2) - 0.3) < 1e-9)
```

```python
# Conversions explicites
print(int("42") + 1, float("3.5"), str(42) + "€")
```

```python
# int() TRONQUE, round() arrondit
print(int(3.99))
print(round(3.99))
```

```python
# Truthy / Falsy : 0, 0.0, "", [], {}, None sont faux, le reste est vrai
print(int(True), bool(0), bool("non vide"), bool([]))
```

```python
# None : l'absence de valeur, à tester avec `is`
valeur = None
print(valeur is None, type(valeur))
```

```python
# isinstance() : bool est un sous-type de int
print(isinstance(3, int), isinstance(True, int))
```

## 5. Afficher : `print()` et f-strings

`print(*objets, sep=" ", end="\n")`. Pour formater, on utilise les **f-strings** : `f"{expression:format}"`.

```python
# Les paramètres sep et end de print()
print("a", "b", "c", sep=" | ", end=".\n")
```

```python
# f-string : insérer des variables dans une chaîne
nom, age, solde = "Alice", 30, 1234.5678
print(f"{nom} a {age} ans")
```

```python
# Nombre de décimales et séparateur de milliers
print(f"{solde:.2f}")
print(f"{solde:,.2f}")
```

```python
# Pourcentage et zéros devant
print(f"{0.256:.1%}")
print(f"{age:05d}")
```

```python
# Alignement : droite, gauche, centré
print(f"{nom:>10}|{nom:<10}|{nom:^10}|")
```

```python
# Bases : binaire, octal, hexadécimal
print(f"{255:b} {255:o} {255:x}")
```

```python
# Debug avec = et accolades littérales
print(f"{age=}")
print(f"{{accolades littérales}}")
```

```python
# repr() vs str() : repr est non ambigu (utile en debug)
s = "ligne 1\nligne 2"
print(str(s))
print(repr(s))
```

```python
# Le même résultat que repr() dans une f-string avec !r
print(f"{s!r}")
```

## 6. Saisir : `input()`

`input()` retourne **toujours** une `str`. Convertir ensuite, et gérer l'erreur.

```python
# Décommente pour tester en interactif :
reponse = input("Ton âge ? ")
# reponse = "30"

try:
    age = int(reponse)
    print(f"Dans 10 ans : {age + 10}")
except ValueError:
    print("Ce n'est pas un entier.")
```

## ⚠️ Pièges

- `7 / 2` donne `3.5` ; `7 // 2` donne `3`.
- `0.1 + 0.2 != 0.3` : comparer les flottants avec une tolérance, ou utiliser `decimal` pour l'argent.
- `int(3.99)` donne `3` (troncature), `round(3.99)` donne `4`.
- `input()` renvoie une chaîne : `"3" + 1` lève une `TypeError`.
- Mélanger tabulations et espaces lève une `IndentationError` : configurer l'éditeur sur 4 espaces.
- Ne pas nommer une variable `list`, `str`, `sum`, `print` : cela masque la fonction native.
